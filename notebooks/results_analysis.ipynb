# 📊 Project – Multi-Criteria Decision Making Applied to Football Recruitment

## 🎯 Context
At the end of the 2023/2024 season, **Stade Brestois 29** finished 3rd in Ligue 1, securing a historic qualification for the **2024/2025 UEFA Champions League**.  
To strengthen the squad for the increased number of matches and maintain competitive performance, the club aims to recruit a **high-level striker**.

Two key decision-makers are involved in this recruitment process:
- **Éric Roy (Head Coach)**: focused on sporting performance and tactical fit  
- **Denis Le Saint (President)**: attentive to financial balance, club image, and the long-term sustainability of the project

## 📝 Objective
The objective of this project is to develop a **Multi-Criteria Decision Making (MCDM)** tool capable of identifying striker profiles best aligned with the combined expectations of both decision-makers.

We apply the **MAVT (Multi-Attribute Value Theory)** methodology to:
- Define criteria relevant to each decision-maker (sporting, financial, psychological, reputational)
- Weight these criteria according to their respective priorities
- Evaluate and rank potential candidates to provide realistic recommendations

## 📂 Data
The data comes from the public *Football Manager 2023* dataset (Kaggle: [Football Manager 2023 Dataset](https://www.kaggle.com/datasets/platinum22/foot-ball-manager-2023-dataset)), including **8,452 players** and **98 variables** (sporting attributes, financial indicators, personal data).

## 🔎 Problem Statement
The goal is not to simply choose **one ideal player per decision-maker**, but to:
1. Generate two rankings (coach-oriented & president-oriented)  
2. Identify **compromise profiles** performing well under both priority systems  

This project illustrates how data science and multicriteria decision support methods can contribute to **strategic recruitment choices** within a professional football club.


## Method used
### 🔽 Initial Filtering of the Dataset
Several preliminary filters are applied to reduce the number of candidate players:

- **Central strikers only** → position targeted by the club  
- **Market value < €15M and wage < €50K/month** → aligned with Stade Brestois’ historically cautious recruitment approach  
- **Age between 20 and 35** → balance between youth (development, resale potential) and experience at high competition level  

---

### 🧑‍🏫 Decision-Maker 1: Head Coach
The head coach prioritizes **sporting performance**: he is looking for a striker who is efficient in front of goal (finishing), physically strong, and dominant in aerial play.  
He favors a **technical, collective, and loyal** player who can commit to a long-term project.  
He also wants to minimize injury risks and values **ambition**, **youth**, and the ability to sustain high-intensity play.  

👉 These are the **key criteria we will evaluate** for this decision-maker.

---

### 👔 Decision-Maker 2: Club President
The president focuses on **financial and institutional aspects**: he wants a player with an affordable transfer fee and salary, while also improving the club’s image.  
He values extra-sporting attributes such as **reputation, fair play, and emotional control**, and also seeks loyalty and ambition.  
Sporting ability remains relevant, but is secondary to economic viability and brand-building considerations.  

👉 These are therefore the **criteria to be evaluated** for this decision-maker.

---

### ⚽ Multi-Criteria Evaluation Methodology (MAVT)

To evaluate the candidate players, we follow a three-step approach:

#### 1. **Criterion Scaling**
Each criterion (age, finishing, heading, injury proneness, etc.) is normalized using a *QuantitativeScale*, with a direction of preference (*maximize* or *minimize*).  
This ensures that criteria expressed in different units (years, ratings, salary amounts...) are made directly comparable.

#### 2. **Value Functions**
Raw performance indicators are transformed into **preference values** using piecewise linear value functions.  
These functions reflect each decision-maker’s priorities:  
For example, younger players receive higher preference, while injury-prone players are penalized.  
The curves visually represent the relationship between performance level and satisfaction gained from that performance.

#### 3. **Weighted Sum Aggregation**
Each criterion receives a weight representing its relative importance.  
Preference values are then aggregated using the **MAVT weighted sum method**, producing a **global score** for each player.  
The final ranking highlights the players who best meet the decision-maker’s strategic goals.

---

### ⚖️ Final Decision – Multi-Decision Maker Compromise

To select the strikers for the 2024–2025 season, we applied a **multi-decision maker compromise method**:

1. We calculated the **MAVT scores** for each player according to the criteria and weightings defined by both **the head coach** and **the president**.  
2. Instead of selecting one player per decision-maker, we **summed the two scores** obtained for each player across the two preference models.  
3. Players with the **highest combined scores** were retained as final recommendations, ensuring a balanced compromise satisfying both decision-makers.

This approach ensures that the final recommendations reflect both **sporting performance requirements** and the club’s **economic and reputational constraints**.
